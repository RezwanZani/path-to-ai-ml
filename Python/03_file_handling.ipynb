# File I/O
Python can be used to perdorm operations (read and write) on a file.

#### Types of all files
- **Text Files**: .txt, .docx, .log etc.
- **Binary Files**: .mp4, .png, .pdf, etc.

### Open
We have to open file before reading or writing

#### Modes
| Character | Meaning |
| --- | --- |
| 'r' | open for reading (default)|
| 'w' | open for writing, truncating the file first|
| 'x' | create a new file and open it for writing|
| 'a' | open for writing, appending to the end of the file if it exists|
| 'b' | binary mode|
| 't' | text mode (default)|
| '+' | open a disk file for updating (reading and writing)|

### Read

````python
## Read the entire file

f = open("03_demo.txt", "r")        # Open a file in read ("r") mode
data = f.read()                     # Read the entire file and return it as a string
print(data)
f.close()
````

````python
## Read specific number of characters

f = open("03_demo.txt", "r")             # Open a file in read ("r") mode
data = f.read(5)                         # Read the first 5 characters of the file and return it as a string
print(data)
f.close()
````

````python
## Read the file line by line

f = open("03_demo.txt", "r")             # Open a file in read ("r") mode

line1 = f.readline()                       # Read the first line of the file and return it as a string
line2 = f.readline()                       # Read the second line of the file and return it as a string
line3 = f.readline()                       # Read the third line of the file and return it as a string
line4 = f.readline()                       # Return an empty string if the end of the file has been reached

print(line1)
print(line2)
print(line3)
print(line4)

f.close()
````

### Write

````python
f = open("03_demo.txt", "w")             # Open a file in write ("w") mode
f.write("This is a new line.\n")        # Write a string to the file
f.write("This is another new line.\n")  # Write another string to the file
f.close()
````

````python
## Apend to a file
f = open("03_demo.txt", "a")             # Open a file in append ("a") mode
f.write("\nThis line will be appended to the file.\n")  # Append a string
f.close()
````

````python
## "r+" mode overwrites the file from the beginning, so it is not suitable for appending data to the end of the file. If you want to append data to the end of the file, you should use "a" mode instead.
f = open("03_demo.txt", "r+")             # Open a file in read and write ("r+") mode
f.write("This line will overwrite the beginning of the file.\n")  # Overwrite the beginning of the file
f.close()
````

#### `with` Syntax

````python
with open("03_demo.txt", "r") as f:
    data = f.read()
    print(data)

# With close the file automatically after the block of code is executed, even if an exception is raised. This is a good practice to ensure that files are properly closed and resources are released.

````

### Keyword Arguments

```python
## Keyword Arguments
def substract(a, b):
    """Returns the difference of two numbers."""
    print(f'{a} - {b} = {a - b}')
    return a - b

substract(5, 10)        # Positional Arguments
substract(b=5, a=10)    # Keyword Arguments
```

```python
def any_functon(*args, **kwargs):
    print('Positional Arguments:', args)
    print('Keyword Arguments:', kwargs)

    for key, values in kwargs.items():
        print(f'{key}: {values}')

any_functon(1, 2, 3, name='Alice', age=30)
```

## Iterator and Generator

```python
## Iterator

s = {10, 20, 30, 40, 50}

# print(s[1])         # This will raise an error because sets are unordered and do not support indexing.

s_iter = iter(s)  # Create an iterator for the set
print(next(s_iter))  # Get the first element from the iterator
print(next(s_iter))  # Get the second element from the iterator
```

```python
l = [x for x in range(500)]          # Example Dataset

def dataloader(chunk_size, data):
    """" Yields chunks of data from the provided dataset."""
    for i in range(0, len(data), chunk_size):
        yield data[i: i + chunk_size]

x = dataloader(10, l)
print(x)                # This will print the generator object, not the data itself.
print(next(x))          # Get the first chunk of data
print(next(x))          # Get the second chunk of data
```

## Lambda Function

```python
import pandas as pd

df = pd.DataFrame({'Temp (F)': [32, 60, 75, 100, 118, 95, 44]})
df

df['Temp (C)'] = df['Temp (F)'].apply(lambda x: (x - 32) * 5.0/9.0)
df
```

```python
df = pd.DataFrame({'A': [1, 5, 6, 7, 11, 15, 20], 'B': [2, 7, 6, 8, 31, 12, 14]})
df

df['A^2'] = df['A'].map(lambda x: x ** 2)
df

df['is_even(B)'] = df['B'].apply(lambda x: 'Even' if x % 2 == 0 else 'Odd')
df
```

### Filter

```python
numbers = [10, 25, 30, 45, 50]

# Filter numbers greater than 25
large_numbers = list(filter(lambda x: x > 25, numbers))
print(large_numbers)  # Output: [30, 45, 50]
```

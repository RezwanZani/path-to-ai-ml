```python
# Mini Calculator
a = float(input("Enter a number: "))
b = float(input("Enter another number: "))
operator = input("Enter operator (+, -, *, /, **):")

if operator == "+":
    print(a + b)
elif operator == "-":
    print(a - b)
elif operator == "*":
    print(a * b)
elif operator == "/":
    print(a / b)
elif operator == "**":
    print(a ** b)

```

```python
# While Loop
counter = 1;

while counter <= 5:
    print(counter * "*")
    counter += 1
```

```python
# List
marks = [23, 45, 34, 39, 41]
print(marks)

# Length
print(len(marks))

# Index
print(marks[0])
print(marks[2])
print(marks[-1])

# Slice
print(marks[0:3])  # [start:end]
print(marks[-3:-1])
print(marks[:])

# Travers
for score in marks:
    print(score)

# Append (add at the end)
marks.append(50)
print(marks)

# Insert (add at a specific location)
marks.insert(1, 48)     # (position, value)
print(marks)

# Clear
marks.clear()
print(marks)        # []
```

```python
# Tuple - Immutable
marks = (98, 96, 76, 65, 88, 96)
print(marks)

print(marks[3])             #65
print(marks.index(65))      # 3
```

```python
# Set
marks = {96, 92, 98, 92, 95, 95, 87, 90, 92}

print(marks, "Length:", len(marks))         # 6 (no repeated value counts)

for score in marks:
    print(score)
```

```python
# Dictionary (key-value pair)
marks = {
    "Math": 136,
    "Physics": 184,
    "Chemistry": 184
}

print("Marks:", marks)
print("Type:", type(marks))

marks["English"] = 164
print("Marks:", marks)

print(marks["Physics"])

for key in marks:
    print(key, marks[key])

```

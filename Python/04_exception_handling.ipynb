## Exception Handling

#### Division by Zero

```python
n = int(input("Enter a number: "))

try:
    a = 10 / n
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

print("Hello World!")
```

#### Name Error

```python
try:
    x = y
except NameError:
    print("Error: Variable 'y' is not defined.")
```

```python
# All exceptions are derived from the base class Exception. You can catch multiple exceptions in a single except block by specifying a tuple of exception types. For example:

try:
    n = int(input("Enter a number: "))
    a = 10 / n
    x = y
except Exception as e:
    print(f"Error: {e}")
```

```python

y = 5  # Define y to avoid NameError

try:                            # Code that may raise an exception is placed inside the try block. If an exception occurs, the code in the except block will be executed.
    n = int(input("Enter a number: "))
    a = 10 / n
    x = y
except Exception as e:          # This will catch any exception that occurs in the try block and store it in the variable e. You can then use e to access information about the exception, such as its type and message.
    print(f"Error: {e}")
else:                           # The else block is executed if no exceptions occur in the try block. This is useful for code that should only run if the try block was successful.  
    print(f"Result: {a}")
    print(f"Value of x: {x}")
finally:                        # The finally block is executed regardless of whether an exception occurred or not. This is useful for cleanup code that should always run, such as closing files or releasing resources.
    print("Execution completed.")
```

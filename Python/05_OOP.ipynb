### `__init__` Function (constructor)
---
All classes have a function called `--init--()`, which is always executed when the class is being initiated.

### Methods
---
Methods are functions that belongs to objects

```python
# Creating class
class Student:
    college_name = "ABCD college"
    def __init__(self, fullname, marks):
        self.name = fullname
        self.marks = marks

    def welcome(self):
        print("Welcome student,", self.name)

    def get_marks(self):
        return self.marks




# Creating object (instance)
s1 = Student("Rezwan", 80)
print(s1.name, s1.marks, s1.college_name)

s1.welcome()
print("marks:", s1.get_marks())
```

#### Static Method
---
Methods that don't use the `self` parameter (work at class level)

> Decorators allow us to wrap another function in order to extend the behaviour of the wrapped function, without permanently modifying it.

```python
# Creating class
class Student:
    college_name = "ABCD college"
    def __init__(self, fullname, marks):
        self.name = fullname
        self.marks = marks

    @staticmethod
    def hello():
        print("Hello, World!")

    def welcome(self):
        print("Welcome student,", self.name)

    def get_marks(self):
        return self.marks

    def get_avg(self):
        sum = 0
        for mark in self.marks:
            sum += mark

        return sum / len(self.marks)


# Creating object (instance)
s1 = Student("Rezwan", [80])
print(f"Name: {s1.name}\nMarks: {s1.marks},\nCollege: {s1.college_name},\nAverage {s1.get_avg()}")

s1.hello()
Student.hello()

```

## Private(like) attributes & methods

**Conceptual Implementations in Python**

Private attributes & methods are meant to be used only within the class and are not accessible from outside the class.

```python
class Account:
    def __init__(self, acc_no, acc_pass):
        self.acc_no = acc_no
        self.acc_pass = acc_pass

acc1 = Account("121212", "password")
print(f"A/C No.: {acc1.acc_no}\nPassword: {acc1.acc_pass}")
```

```python
# Password should be hidden. It should be inaccessible.

class Account:
    def __init__(self, acc_no, acc_pass):
        self.acc_no = acc_no
        self.__acc_pass = acc_pass          # Making the variable private (inaccessible outside the class)

    def show_pass(self):
        print("Password:", self.__acc_pass) # Accessible within class

acc1 = Account("121212", "password")

try:
    print(f"A/C No.: {acc1.acc_no}\nPassword: {acc1.__acc_pass}")
except Exception as e:
    print(e)

acc1.show_pass()                            # Accessed
```

```python
# Private Method

class Person():
    def __init__(self, name):
        self.__name = name

    def __hello(self):
        print("Hello")

    def welcome(self):
        print(f"Hello, {self.__name}!")

p1 = Person("Rezwan")

try:
    print(p1.__hello())
except Exception as e:
    print(e)

p1.welcome()
```

## Inheritance

When one class(child/derived) derives the properties & methods of another class(parent/base).

```python
class Car:
    @staticmethod
    def start():
        print("Car started...")

    @staticmethod
    def stop():
        print("Car stopped.")

class ToyotaCar(Car):
    def __init__(self, name):
        self.name = name

car1 = ToyotaCar("Fortuner")
car2 = ToyotaCar("Corolla")

car1.start()
car2.stop()
```

```python
# Multiple Inheritance
class Car:
    @staticmethod
    def start():
        print("Car started...")

    @staticmethod
    def stop():
        print("Car stopped.")

class ToyotaCar(Car):
    def __init__(self, brand):
        self.brand = brand

class Fortuner(ToyotaCar):
    def __init__(self, model):
        self.model = model

car1 = Fortuner("Diesel")
car1.start()
car1.stop()
```

```python
# Multiple Inheritance

# Parent Class 1
class Flyer:
    def fly(self):
        print("Flying in the sky")

# Parent Class 2
class Swimmer:
    def swim(self):
        print("Swimming in the water")

# Child Class inheriting from both Flyer and Swimmer
class Duck(Flyer, Swimmer):
    def quack(self):
        print("Quack quack!")

# Example usage
donald = Duck()
donald.fly()    
donald.swim()   
donald.quack()  
```

#### The super() Method in Python
The `super()` function allows a child class to call methods from its parent (or superclass), commonly used to invoke the parent class constructor (`__init__`) and reuse initialization logic.

```python
class Animal:
    def __init__(self, name: str, species: str):
        self.name = name
        self.species = species

    def make_sound(self):
        print(f"{self.name} makes a sound.")


class Dog(Animal):
    def __init__(self, name: str, breed: str):
        # Call the parent constructor using super()
        super().__init__(name=name, species="Canine")
        self.breed = breed

    def make_sound(self):
        # Extend the parent method behavior
        super().make_sound()
        print(f"{self.name} barks: Woof! Woof!")


# Usage
dog = Dog(name="Buddy", breed="Golden Retriever")

print(f"Name: {dog.name}")
print(f"Species: {dog.species}")
print(f"Breed: {dog.breed}")

dog.make_sound()
```

### Class Methods in Python
A class method is a method bound to the class itself rather than an instance of the class. It receives the class (`cls`) as its first implicit argument instead of the instance (`self`).

_Class methods are marked using the `@classmethod` decorator._

Key Characteristics
- Bound to Class: Modifies or accesses class-level state rather than instance-specific attributes.
- First Argument (cls): Automatically passes the class object as the first parameter.
- Alternative Constructors: Commonly used as factory methods to create class instances from different formats (e.g., parsing strings, JSON, or timestamps).

```python
class Person:
    # Class attribute
    species = "Homo sapiens"

    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    # Class method as an alternative constructor / factory method
    @classmethod
    def from_birth_year(cls, name: str, birth_year: int, current_year: int = 2026):
        calculated_age = current_year - birth_year
        return cls(name, calculated_age)

    # Class method accessing/modifying class-level data
    @classmethod
    def change_species(cls, new_species: str):
        cls.species = new_species


# Standard instantiation
p1 = Person("Alice", 30)

# Instantiation using the class method (factory)
p2 = Person.from_birth_year("Bob", 1996)

print(p1.name, p1.age)        # Output: Alice 30
print(p2.name, p2.age)        # Output: Bob 30

# Modifying class attribute via class method
Person.change_species("Homo sapiens sapiens")
print(p1.species)             # Output: Homo sapiens sapiens
print(p2.species)             # Output: Homo sapiens sapiens
```

## Property Decorator in Python

The `@property` decorator is a built-in feature that allows a method to be accessed like an attribute, without calling it explicitly with parentheses `()`.

It is primarily used for **getters**, **setters**, and **deleters**, enabling data validation and encapsulation while keeping a clean, attribute-like syntax.

---

### Key Benefits

* **Clean Syntax:** Access attributes via `obj.value` instead of calling getter methods like `obj.get_value()`.
* **Encapsulation & Validation:** Add logic and validation rules before reading, modifying, or deleting an attribute.
* **Backward Compatibility:** Change an internal attribute into a computed or validated value without breaking existing code that accesses it directly.

```python
class Temperature:
    def __init__(self, celsius: float):
        self._celsius = celsius  # Protected attribute

    # Getter: access as temp.celsius
    @property
    def celsius(self) -> float:
        return self._celsius

    # Setter: runs on temp.celsius = value
    @celsius.setter
    def celsius(self, value: float):
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is impossible!")
        self._celsius = value

    # Computed read-only property
    @property
    def fahrenheit(self) -> float:
        return (self._celsius * 9 / 5) + 32

    # Deleter: runs on del temp.celsius
    @celsius.deleter
    def celsius(self):
        print("Resetting celsius temperature.")
        del self._celsius


# Usage
t = Temperature(25)

# Accessing properties like regular attributes
print(f"Celsius: {t.celsius}°C")
print(f"Fahrenheit: {t.fahrenheit}°F")

# Modifying using the setter
t.celsius = 30
print(f"Updated Fahrenheit: {t.fahrenheit}°F")

# Attempting invalid value triggers validation error
try:
    t.celsius = -300
except ValueError as e:
    print(f"Error: {e}")

# Deleting the attribute
del t.celsius
```

## Polymorphism : Operator Overloading

When the same operator is allowed to have different meaning according to the context.

### Operators & Dunder functions

| Expression | Description | Equivalent Dunder Call |
| :--- | :--- | :--- |
| `a + b` | #addition | `a.__add__(b)` |
| `a - b` | #subtraction | `a.__sub__(b)` |
| `a * b` | #multiplication | `a.__mul__(b)` |
| `a / b` | #division | `a.__truediv__(b)` |
| `a % b` | #modulo | `a.__mod__(b)` |

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __sub__(self, other):
        return Number(self.value - other.value)

    def __mul__(self, other):
        return Number(self.value * other.value)

    def __truediv__(self, other):
        return Number(self.value / other.value)

    def __floordiv__(self, other):
        return Number(self.value // other.value)

    def __mod__(self, other):
        return Number(self.value % other.value)

    def __eq__(self, other):
        return self.value == other.value

    def __lt__(self, other):
        return self.value < other.value

    def __gt__(self, other):
        return self.value > other.value

    def __repr__(self):
        return f"Number({self.value})"


a = Number(10)
b = Number(3)

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a == b)
print(a < b)
print(a > b)
```

```python
class Vector2D:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    # Overload '+' operator
    def __add__(self, other):
        if isinstance(other, Vector2D):
            return Vector2D(self.x + other.x, self.y + other.y)
        return NotImplemented

    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"


v1 = Vector2D(2, 4)
v2 = Vector2D(3, 1)

# Equivalent to v1.__add__(v2)
v3 = v1 + v2
print(v3)  # Output: Vector2D(5, 5)
```

```python
class Student:
    def __init__(self, name: str, marks: int):
        self.name = name
        self.marks = marks

    # Overload '<' operator
    def __lt__(self, other):
        return self.marks < other.marks

    # Overload '==' operator
    def __eq__(self, other):
        return self.marks == other.marks


alice = Student("Alice", 85)
bob = Student("Bob", 92)

print(alice < bob)   # Output: True (invokes alice.__lt__(bob))
print(alice == bob)  # Output: False (invokes alice.__eq__(bob))
```

## 0. Starter Code

```python
import numpy as np
import pandas as pd
np.random.seed(24)
num_employees = 50 # you can increase this if needed
# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)
# Fixed set of roles (repetition allowed)
roles_list = [
"HR Manager",
"Software Developer",
"Accountant",
"Marketing Executive",
"Sales Executive",
"Backend Developer",
"HR Executive",
"Financial Analyst",
"Digital Marketer",
"Operations Manager"
]
emp_roles = np.random.choice(roles_list, num_employees)
# Create DataFrame
data = pd.DataFrame({
"emp_id": emp_ids,
"emp_age": emp_ages,
"emp_salary": emp_income,
"role": emp_roles
})
# Save dataset
data.to_csv("staff_data.csv", index=False)
```

## Q1

```python
# Make the arrays
ids, ages, salaries = np.genfromtxt('staff_data.csv',
                                        delimiter=",",
                                        skip_header=1,
                                        usecols=(0, 1, 2),
                                        unpack=True)

#age
print("emp_ages: ", ages)
print(f"Data Type: {ages.dtype} | No. of Dimension: {ages.ndim}| Shape: {ages.shape} | Total Elements: {ages.size}")
print("\n----------------------------------------------------------------------------------------------------------\n")

# income
print("emp_salaries: ", salaries)
print(f"Data Type: {salaries.dtype} | No. of Dimension: {salaries.ndim}| Shape: {salaries.shape} | Total Elements: {salaries.size}")
```

## Q2

```python
max_salary = np.max(salaries)
min_salary = np.min(salaries)
avg_salary = np.mean(salaries)
print(f"\nMaximum Salary: {max_salary} | Minimum Salary: {min_salary} | Average Salary: {avg_salary}")
```

## Q3

```python
condition = (ages >= 25) & (ages <= 40) & (salaries > 50000)

filtered_salaries = salaries[condition]
filtered_ages = ages[condition]

filtered_ids = ids[condition]


print("\nFiltered Salaries: ", filtered_salaries)
print("Filtered Ages: ", filtered_ages)
print("Total employees satisfied the condition: ", filtered_ids.size)
```

## Q4

```python
avg_age = np.mean(ages)

new_ages = np.where(ages < avg_age, ages + 4, ages - 2)

print("\nOriginal Ages: ", ages)
print("\nNew Ages: ", new_ages)
print("\nAverage Age: ", avg_age)
```

## Q5

```python
roles = np.genfromtxt('staff_data.csv',
                    delimiter=",",
                    skip_header=1,
                    usecols=(3),
                    dtype=str)

tech_roles = ["Software Developer", "Backend Developer"]
is_tech = np.isin(roles, tech_roles)

new_salaries = np.where(is_tech, salaries + (salaries * 0.08), salaries)

old_total_salary = np.sum(salaries)
new_total_salary = np.sum(new_salaries)
salary_increase = new_total_salary - old_total_salary

print("\nOld Total Salary: ", old_total_salary)
print("\nNew Total Salary: ", new_total_salary)
print("\nTotal Salary Increase: ", salary_increase)
```

```python
emp_data['previous_salary'] = emp_data['emp_salary']

mask = emp_data['role'].isin(['Software Developer', 'Backend Developer'])

emp_data.loc[mask, 'new_salary'] = emp_data.loc[mask, 'previous_salary'] * 1.12
emp_data.loc[~mask, 'new_salary'] = emp_data.loc[~mask, 'previous_salary'] * 1.05

emp_data.loc[:, 'increase_of_salar'] = emp_data['new_salary'] - emp_data['previous_salary']

print(emp_data[['role', 'previous_salary', 'new_salary', 'increase_of_salar']])
```

#### Solution to **Question - 1**

```python
num_str = input("Enter space-separated integers:")
num_str = num_str.split()

num_list = [int(x) for x in num_str]    # Convert to number

output = []

for num in num_list:
    if num % 4 == 0 and num % 8 != 0:
        output.append(num)

print(output)
```

#### Solution to **Question - 2**

```python
def compute_ratio(x, y):
    try:
        result = (x + y) / (x * y)
    except ZeroDivisionError:
        return "Error: Divison by zero is not possible"
    else:
        return result

print(compute_ratio(5, 1))
```

#### Solution to **Question - 3**

```python
f = open("demo.txt", "r")
data = f.read()

words = data.strip(".").split()

longest_word = ""

for word in words:
    if len(word) > len(longest_word):
        longest_word = word

print(longest_word)
```

#### Solution to **Question - 4**

```python
sentence = input("Enter any sentence:")
words = sentence.strip(".").split()

vowels = ['a', 'e', 'i', 'o', 'u']

output = []

for word in words:
    if word[0].lower() in vowels:
        output.append(word)

print(output)
```

#### Solution to **Question - 5**

```python
word = input("Enter any word:")

left = 0
right = len(word) - 1

palindrome = True

while left <= right:
    if word[left].lower() == word[right].lower():
        left += 1
        right -= 1
    else:
        palindrome = False
        break

if palindrome:
    print("Palindrome")
else:
    print("Not Palindrome")
```
